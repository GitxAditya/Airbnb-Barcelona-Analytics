```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt

data_path = '../02_Data/'
datasets = {
    'listings': pd.read_csv(os.path.join(data_path, 'listings.csv')),
    'calendar': pd.read_csv(os.path.join(data_path, 'calendar.csv')),
    'reviews': pd.read_csv(os.path.join(data_path, 'reviews.csv')),
    'neighbourhoods': pd.read_csv(os.path.join(data_path, 'neighbourhoods.csv'))
}
```

```python
listings = datasets['listings']
calendar = datasets['calendar']
reviews = datasets['reviews']
neighbourhoods = datasets['neighbourhoods']

# Task 12: Unique Identifiers and Key Integrity
print("--- Key Integrity ---")
print(f"Unique listing IDs in listings: {listings['id'].nunique():,}")
print(f"Unique listing IDs in calendar: {calendar['listing_id'].nunique():,}")
print(f"Unique listing IDs in reviews: {reviews['listing_id'].nunique():,}")
print(f"Unique review IDs in reviews: {reviews['id'].nunique():,}")

# Orphan check — do all calendar/review IDs exist in listings?
cal_orphans = set(calendar['listing_id']) - set(listings['id'])
rev_orphans = set(reviews['listing_id']) - set(listings['id'])
print(f"\nCalendar listing IDs missing from listings: {len(cal_orphans):,}")
print(f"Review listing IDs missing from listings: {len(rev_orphans):,}")

# Task 13: Neighbourhood coverage
print("\n--- Neighbourhood Coverage ---")
print(f"Unique neighbourhoods in listings: {listings['neighbourhood'].nunique()}")
print(f"Unique neighbourhoods in reference: {neighbourhoods['neighbourhood'].nunique()}")
print(f"Listings with missing neighbourhood: {listings['neighbourhood'].isnull().sum()}")

unmatched = set(listings['neighbourhood'].dropna()) - set(neighbourhoods['neighbourhood'])
print(f"Neighbourhoods in listings not in reference: {len(unmatched)}")
if unmatched:
    print(f"Examples: {list(unmatched)[:5]}")

# Latitude / Longitude ranges
print(f"\nLatitude range: {listings['latitude'].min():.4f} to {listings['latitude'].max():.4f}")
print(f"Longitude range: {listings['longitude'].min():.4f} to {listings['longitude'].max():.4f}")

# Task 14: Join risk — does joining inflate data?
sample = listings[['id', 'price']].head(1)
sample_id = sample['id'].iloc[0]
cal_rows = (calendar['listing_id'] == sample_id).sum()
rev_rows = (reviews['listing_id'] == sample_id).sum()
print(f"\n--- Join Risk Test (listing {sample_id}) ---")
print(f"Calendar rows for this listing: {cal_rows:,} (would multiply by {cal_rows}x)")
print(f"Review rows for this listing: {rev_rows:,} (would multiply by {rev_rows}x)")
```
