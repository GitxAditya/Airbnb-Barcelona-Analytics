```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt

data_path = '../02_Data/'
datasets = {
    'listings': pd.read_csv(os.path.join(data_path, 'listings.csv')),
    'calendar': pd.read_csv(os.path.join(data_path, 'calendar.csv')),
    'reviews': pd.read_csv(os.path.join(data_path, 'reviews.csv')),
    'neighbourhoods': pd.read_csv(os.path.join(data_path, 'neighbourhoods.csv'))
}
```

```python
calendar = datasets['calendar'].copy()
reviews = datasets['reviews'].copy()

calendar['date'] = pd.to_datetime(calendar['date'], errors='coerce')
reviews['date'] = pd.to_datetime(reviews['date'], errors='coerce')

print("--- Calendar Date Coverage ---")
print(f"Min date: {calendar['date'].min()}")
print(f"Max date: {calendar['date'].max()}")
print(f"Distinct months: {calendar['date'].dt.to_period('M').nunique()}")
print(f"Distinct years: {calendar['date'].dt.year.nunique()}")

print("\n--- Reviews Date Coverage ---")
print(f"Min date: {reviews['date'].min()}")
print(f"Max date: {reviews['date'].max()}")
print(f"Distinct months: {reviews['date'].dt.to_period('M').nunique()}")
print(f"Distinct years: {reviews['date'].dt.year.nunique()}")

# Missing months?
cal_months = set(calendar['date'].dt.to_period('M').unique())
expected = pd.period_range(calendar['date'].min(), calendar['date'].max(), freq='M')
print(f"\nMissing months in calendar: {set(expected) - cal_months}")
```
