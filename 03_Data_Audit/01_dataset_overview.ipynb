```python
import pandas as pd
import numpy as np
import os

data_path = '../02_Data/'

# Load the datasets
try:
    listings = pd.read_csv(os.path.join(data_path, 'listings.csv'))
    print("listings.csv loaded successfully.")
except Exception as e:
    print(f"Error loading listings.csv: {e}")

try:
    calendar = pd.read_csv(os.path.join(data_path, 'calendar.csv'))
    print("calendar.csv loaded successfully.")
except Exception as e:
    print(f"Error loading calendar.csv: {e}")

try:
    neighbourhoods = pd.read_csv(os.path.join(data_path, 'neighbourhoods.csv'))
    print("neighbourhoods.csv loaded successfully.")
except Exception as e:
    print(f"Error loading neighbourhoods.csv: {e}")

try:
    reviews = pd.read_csv(os.path.join(data_path, 'reviews.csv'))
    print("reviews.csv loaded successfully.")
except Exception as e:
    print(f"Error loading reviews.csv: {e}")
```

```python
# Create a summary dataframe for dimensions
datasets = {
    'listings': listings,
    'calendar': calendar,
    'neighbourhoods': neighbourhoods,
    'reviews': reviews
}

dimensions = []
for name, df in datasets.items():
    dimensions.append({
        'Dataset': name,
        'Rows': df.shape[0],
        'Columns': df.shape[1]
    })

df_dimensions = pd.DataFrame(dimensions)
print("--- Dataset Dimensions ---")
display(df_dimensions)
```

```python
# Loop through each dataset to display info and head
for name, df in datasets.items():
    print(f"\n{'='*20}")
    print(f" DATASET: {name.upper()}")
    print(f"{'='*50}")
    
    print("\n--- Data Types & Non-Null Counts ---")
    df.info()
    
    print("\n--- First 3 Rows ---")
    display(df.head(3))
    
    print("\n--- Summary Statistics (Numeric Columns) ---")
    display(df.describe())
```

```python
print(f"Listings - Total rows: {len(listings)}, Unique IDs: {listings['id'].nunique()}")

# Calendar:
calendar_grain = calendar.groupby(['listing_id', 'date']).size().reset_index(name='count')
print(f"Calendar - Total rows: {len(calendar)}, Unique listing-date combos: {len(calendar_grain)}")

# Reviews:
print(f"Reviews - Total rows: {len(reviews)}, Unique Review IDs: {reviews['id'].nunique()}")

# Neighbourhoods:
print(f"Neighbourhoods - Total rows: {len(neighbourhoods)}, Unique Neighbourhoods: {neighbourhoods['neighbourhood'].nunique()}")
```
