```python
import pandas as pd
import numpy as np
import os

data_path = '../02_Data/'
datasets = {
    'listings': pd.read_csv(os.path.join(data_path, 'listings.csv')),
    'calendar': pd.read_csv(os.path.join(data_path, 'calendar.csv')),
    'reviews': pd.read_csv(os.path.join(data_path, 'reviews.csv')),
    'neighbourhoods': pd.read_csv(os.path.join(data_path, 'neighbourhoods.csv'))
}
```

```python
# 1. Completely duplicated rows
print("--- Completely Duplicated Rows ---")
for name, df in datasets.items():
    dup_count = df.duplicated().sum()
    print(f"{name}: {dup_count:,} fully duplicated rows")

# 2. Candidate unique keys
print("\n--- Candidate Unique Keys ---")
print(f"listings['id'] unique: {datasets['listings']['id'].is_unique}")
print(f"calendar listing-date unique: {datasets['calendar'].groupby(['listing_id','date']).ngroups == len(datasets['calendar'])}")
print(f"reviews['id'] unique: {datasets['reviews']['id'].is_unique}")
print(f"neighbourhoods unique: {datasets['neighbourhoods']['neighbourhood'].is_unique}")

# 3. Check duplicate listing IDs in listings
dup_listings = datasets['listings'][datasets['listings'].duplicated(subset=['id'], keep=False)]
print(f"\nDuplicate listing IDs: {len(dup_listings)}")

# 4. Check repeated listing-date pairs in calendar
cal_dups = datasets['calendar'][datasets['calendar'].duplicated(subset=['listing_id','date'], keep=False)]
print(f"Repeated listing-date pairs in calendar: {len(cal_dups)}")

# 5. Check repeated review IDs
rev_dups = datasets['reviews'][datasets['reviews'].duplicated(subset=['id'], keep=False)]
print(f"Repeated review IDs: {len(rev_dups)}")

# 6. Check if a listing can have multiple reviews (expected)
reviews_per_listing = datasets['reviews'].groupby('listing_id').size()
print(f"\nMax reviews per single listing: {reviews_per_listing.max()}")
print(f"Avg reviews per listing: {reviews_per_listing.mean():.1f}")
```
