```python
import pandas as pd
import numpy as np
import os
import re
import matplotlib.pyplot as plt

RAW_PATH = '../02_Data/'
PROCESSED_PATH = '../02_Data/processed'
os.makedirs(PROCESSED_PATH, exist_ok=True)

def clean_price(series):
    return pd.to_numeric(
        series.astype(str).str.replace(r'[\$,€£]', '', regex=True).str.strip() , errors='coerce'
    )

def snapshot(df, name):
    return {
        'dataset' : name,
        'rows' : df.shape[0],
        'cols': df.shape[1],
        'columns': list(df.columns),
        'dtypes': df.dtypes.astype(str).to_dict()
    }
```

```python
nb_raw = pd.read_csv(os.path.join(RAW_PATH, 'neighbourhoods.csv'))
nb = nb_raw.copy()

print(f"Rows: {len(nb)}, Columns: {list(nb.columns)}")

# duplicates
# You must strip whitespace BEFORE dropping duplicates
nb['neighbourhood'] = nb['neighbourhood'].str.strip()
nb['neighbourhood'] = nb['neighbourhood'].str.title()
print(f"Duplicate neighbourhoods: {nb.duplicated(subset= ["neighbourhood"]).sum()}")

# standardize labels
for cols in nb.columns:
    if nb[cols].dtypes == 'object':
        nb[cols] = nb[cols].astype('str').str.strip().str.title()
```

```python
nb.to_csv(os.path.join(PROCESSED_PATH, 'neighbourhoods_clean.csv'), index=False)
print(f"Saved neighbourhoods_clean.csv  ({nb.shape[0]} rows)")
```
