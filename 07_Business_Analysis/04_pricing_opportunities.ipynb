```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

listings = pd.read_csv('../02_Data/processed/listings_clean.csv', low_memory=False)
```

```python
# 1. Create the benchmark
benchmark = listings.groupby(['neighbourhood', 'room_type'])['price'].median().rename('benchmark_price').reset_index()

# 2. Merge back to main data
pricing_df = listings.merge(benchmark, on=['neighbourhood', 'room_type'], how='left')

# 3. Calculate difference and percentage
pricing_df['price_diff'] = pricing_df['price'] - pricing_df['benchmark_price']
pricing_df['price_diff_pct'] = (pricing_df['price_diff'] / pricing_df['benchmark_price'] * 100).round(1)

# 4. Find extreme over/under-priced listings (Top 10 Overpriced)
print("--- Top 10 Overpriced Listings (vs. Neighbourhood/Room Benchmark) ---")
display(pricing_df.nlargest(10, 'price_diff_pct')[['name', 'neighbourhood', 'room_type', 'price', 'benchmark_price', 'price_diff_pct']])

# 5. Find extreme under-priced listings
print("\n--- Top 10 Underpriced Listings ---")
display(pricing_df.nsmallest(10, 'price_diff_pct')[['name', 'neighbourhood', 'room_type', 'price', 'benchmark_price', 'price_diff_pct']])
```
