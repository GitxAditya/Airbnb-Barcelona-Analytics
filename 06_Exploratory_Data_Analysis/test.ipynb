```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

listings = pd.read_csv('../02_Data/processed/listings_clean.csv', low_memory=False)

print(f"Total listings:         {len(listings):,}")
print(f"Listings with price:    {listings['price'].notna().sum():,}")
print(f"Listings without price: {listings['price'].isna().sum():,} ({listings['price'].isna().mean()*100:.1f}%)")
```

```python
# What are these listings without prices?
no_price = listings[listings['price'].isna()]

print(f"Total listings without price: {len(no_price):,}")
print(f"\nBreakdown by room_type:")
print(no_price['room_type'].value_counts(dropna=False))
print(f"\nBreakdown by property_type:")
print(no_price['property_type'].value_counts(dropna=False).head(10))
print(f"\nBreakdown by neighbourhood (top 10):")
print(no_price['neighbourhood'].value_counts(dropna=False).head(10))
print(f"\nSample of listing names without prices:")
print(no_price[['id', 'name', 'room_type', 'neighbourhood']].head(10))
```
