```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

listings = pd.read_csv('../02_Data/processed/listings_clean.csv')
reviews = pd.read_csv('../02_Data/processed/reviews_clean.csv', parse_dates=['date'])
calendar = pd.read_csv('../02_Data/processed/calendar_clean.csv', parse_dates=['date'])
```

```python
sh_overall = listings.groupby('host_is_superhost').agg(
    listings = ('id','count'),
    median_price = ('price', 'median'),
    pct_entire_home=('room_type', lambda s: round((s=='Entire home/apt').mean()*100,1)),
    median_reviews=('number_of_reviews', 'median')
).round(1)

display(sh_overall)

within_nb = listings.groupby(['neighbourhood', 'host_is_superhost'])['price'].median().unstack()
within_nb['diff'] = within_nb[True] - within_nb[False]
within_nb = within_nb.dropna().sort_values('diff', ascending=False)

print(f"\n Within-Neighbourhood Superhost Premium (median €)")
display(within_nb.head(10))
print(f"\nMedian premium across neighbourhoods: €{within_nb['diff'].median():.0f}")

# Similarly control for room type
within_rt = listings.groupby(['room_type', 'host_is_superhost'])['price'].median().unstack()
print("\n Within-Room-Type Superhost Premium")
display(within_rt)

# Missing status check
print(f"\nListings with missing host status: {listings['host_is_superhost'].isna().sum()}")
```
